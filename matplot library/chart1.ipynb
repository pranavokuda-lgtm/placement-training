```python
import matplotlib.pyplot as plt
import pandas as pd
```

barplot

```python
course=['Python','Java','C++','Data Science','Machine Learning']
enrolled=[200,150,300,250,400]
plt.bar(course,enrolled)
plt.show()
```

```python
course_enroll={
    "courses": ['java', 'python', 'c++', 'data science', 'machine learning'],
    "enrolled": [200, 150, 300, 250, 400]
}
course_df=pd.DataFrame(course_enroll)
#create plot
plt.figure(figsize=(9,5))
bar=plt.bar(course_df['courses'],course_df['enrolled'],
        color=['#EB0800','#EB7D00','#E3EB00','#412EBD','#882EBD'],
        edgecolor='black',width=0.6
       )
#creating title and labels
plt.title('Course Enrolled Students overview')
plt.xlabel('Course details',fontsize=12)
plt.ylabel('course enrolled students',fontsize=12)
plt.bar_label(bar,padding=3,fontweight='bold',fontsize=10)
#improve 
plt.ylim(0,500)
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.gca().spines['top'].set_visible(false)
plt.gca().spines['right'].set_visible(false)
plt.tight_layout()
plt.show()
```

line plot

```python
weekly_sales={
    "days":['Mon','Tue','Wed','Thu','Fri','Sat','Sun'],
    "first_week":[35000,30000,38000,32000,39000,42000,45000],
    "second_week":[32000,35000,42000,37000,34000,43000,48000]
}
sales=pd.DataFrame(weekly_sales)
#sales=pd.data(figsize=(9,5))
#----------------first week sales------------
plt.plot(sales["days"],sales["first_week"],color="red",ls="--",marker="*")
#----------------second week sales------------
plt.plot(sales["days"],sales["second_week"],color="blue",ls="--",marker="o")
plt.title("Shop Weekly Sales Overview",fontsize=15,fontweight="bold")
plt.xlabel("Days of the week",fontsize=12)
plt.ylabel("Sales Amount per day",fontsize=12)
plt.legend(["first week sales","second week sales"],loc="upper left")
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.show()
```

Scatter plot


```python
student_age={
    "sid":['s01','s02','s03','s04','s05','s06','s07','s08','s09','s10','s11'],
    "age": [20, 22, 21, 20, 21, 22, 23, 22, 20, 24, 21]
}
std_df=pd.DataFrame(student_age)
plt.figure(figsize=(9,5))
plt.scatter(std_df['age'],std_df['sid'])
plt.show()
```

```python
student_age= {
    "age":[20,22,21,20,21,22,23,22,20,24,22,21,21,21,20,23,23,24,25,25,32],
    "marks":[490,480,470,460,450,440,430,490,480,470,460,450,440,430,490,480,470,460,450,440,430]
}
std_df=pd.DataFrame(student_age)
plt.figure(figsize=(8,5))
plt.scatter(std_df['age'],std_df['marks'],color='red',marker='o',s=100,edgecolor='yellow')
#create title and labels
plt.title('Student Ages and Marks',fontweight='bold',fontsize=15)
plt.xlabel('Age',fontsize=12)
plt.ylabel('Marks',fontsize=12)
plt.legend(['Student Data'],loc='upper right')
plt.show()
```

pie chart

```python
weekly_sales={
    'Brand':["Oneplus","Samsung","Apple","Vivo","Oppo"],
    'Sales':[23,10,19,32,25]

}
sales_df=pd.DataFrame(weekly_sales)
plt.pie(sales_df['Sales'],labels=sales_df['Brand'],explode=[0,0,0.1,0,0],shadow=True,autopct='%1.1f%%',pctdistance=0.7,startangle=30)
plt.legend(loc='right',bbox_to_anchor=(1.2,1))

plt.show()
```

```python
data=pd.read_excel("expense.xlsx")
exp=pd.DataFrame(data)
exp_group=exp.groupby('Category')['Amount'].sum()
plt.pie(exp_group.values,labels=exp_group.index,explode=[0,0.1,0,0,0],shadow=True,autopct='%1.1f%%',pctdistance=0.7,startangle=30)
plt.legend(loc='right',bbox_to_anchor=(1.2,1))

plt.show()
```

Histogram

```python
temp_celsius=[28,29,30,31,31,30,29,28,27,26,25,24,24,24]
plt.hist(temp_celsius,bins=5,color='green',alpha=0.7,edgecolor='black')
#creating title and labels
plt.title('Distribution of today hourly temperatures',fontsize=15,fontweight='bold',pad=5)
plt.xlabel('Temperature (°C)',fontsize=12)
plt.ylabel('Frequency',fontsize=12)
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.show()
```

```python
# Weekly Sales Data (in units)
weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}
sales=df=pd.DataFrame(weekly_sales)
plt.stackplot(sales['Days'], sales['Week1'], sales['Week2'], sales['Week3'], labels=['Week1', 'Week2', 'Week3'])
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Weekly Sales Data')
plt.legend()
plt.show()
```
