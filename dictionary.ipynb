```python
student={'id':"24cs100",
         'name':"sagar",
         'age':20,
         'dept':"cseb"
         }
#Retrive a single value using key
get_value=student.get('age')
print(get_value)
#check all the keys 
dict_keys=student.keys()
print(dict_keys)
#Retrieve all the values
dict_values=student.values()
print(dict_values)
#Update 
update_age=student.update({"age":21})
print(student)
#set default
dep_val=student.setdefault("dept","cse")
print(dep_val)
new_loc=student.setdefault("place","udupi")
print(student)
#copy
c=student.copy()
print(c)
#pop - remove key-value based on key
student.pop("dept")
print(student)
#popitem
student.popitem()
print(student)
#dict_items
print(student.items())
```

```python
student={'id':"24cs100",
         'name':"sagar",
         'age':20,
         'dept':"cseb"
         }
#Itr over keys
for k in student:
    print(k)
#Itr over values
for v in student.values():
    print(v)
#Itr over key-values 
for k,v in student.items():
    print(k,":",v)
```
