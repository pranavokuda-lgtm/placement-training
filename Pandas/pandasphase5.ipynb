Compare,Pivot & Melting

```python
import pandas as pd
```

```python
shop_7sep={
    "Fruites":['Mango','Apple','Banana','papaya'],
    "price":[100,150,60,30],
    "Qunt":[10,8,12,14]
}
stock_7sep=pd.DataFrame(shop_7sep)

shop_8sep={
    "Fruites":['Mango','Apple','Banana','papaya'],
    "price":[120,200,60,35],
    "Qunt":[15,6,24,17]
}
stock_8sep=pd.DataFrame(shop_8sep)

#compare
stock_7sep.compare(stock_8sep)
stock_7sep.compare(stock_8sep,align_axis=0)
stock_7sep.compare(stock_8sep,keep_shape=True)
stock_7sep.compare(stock_8sep,keep_equal=True)

stock_7sep.compare(stock_8sep,align_axis=0,keep_shape=True,keep_equal=True)
```

Pivoting 

```python
monthly_sales={
    "product":['laptop','mobile','laptop','mobile','laptop','laptop'],
    "month":['jan','jan','feb','feb','mar','mar'],
    "price":[95000,25000,67000,12000,65000,75000]

}
sales=pd.DataFrame(monthly_sales)
#pivot
sales.pivot_table(
    columns='month',
    index='product',
    values='price',
    aggfunc='sum'
)
```

```python
Shop_sales={
    "product":['laptop','mobile','mouse'],
    "jan":[95000,25000,500],
    "feb":[67000,12000,400],
    "mar":[65000,15000,600]
}
sales_df=pd.DataFrame(Shop_sales)
sales_df.melt(id_vars='product', var_name='month', value_name='price')
```
