list indexing and slicing is same as string

```python
#list fuctions 
color=["red","green","black","blue"]

#index  0       1        2      3  
print("org list:",color)
print(len(color))#lengths
color.append("gold")#replace an element
print("after adding gold:",color)
color[1]="White"#replace an element
print("after replacing:",color)
color.insert(2,"pink")#insert value in index position
print("after inserting pink:",color)
color.remove("black")#removing an element
color.pop()
print("after poping",color)
```

```python
#looping
color=["red","green","black","blue"]
#using index ->acces the list element
for i in range (len(color)):
    print(i,color[i])
#directly->acess the list element 
for c in color:
    print(c,end=" ")
    
```

```python
#=============list comprehension==============
#rule -> new_list=[expression for item in iterable if condition
n=10
even=[i for i in range(1,n+1) if i%2==0]
print(even)
m=5
sqr=[x**2 for x in range (1,m+1)]
print(sqr)
```
