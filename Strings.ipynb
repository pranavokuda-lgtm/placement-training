string indexing and slicing

```python
#indexing
str="DARKCHOKLATE"
print("org:",str)
#positive indexing
print("5th:",str[5])
print("last:",str[len(str)-1])
#nagative indexing
print(str[-1])#print last element
print(str[-len(str)])##print first element
```

```python
#slicing -string_name[start:stop:step]
#defoult                0   len(s)  1
str="DARKCHOKLATE"
print(str[0:4])
print(str[:4])
print(str[4:len(str)])
print(str[4:])
print(str[0:len(str):2])
print(str[12:3:-1])
#reverse string
rev_string=str[::-1]
print(rev_string)
```

string functions

```python
s="raju told me that he is coder"
print(s+"it's real")#cocatation 
print(len(s))
print(s.endswith("der"))#is string ends with the substring
print(s.replace("he is","i am"))#replacing old to new
print(s.find("a"))#first index of sub string
print(s.count("e"))#cout the frequency of sub string
print(s.capitalize())#capitalise the first letter of the string
print(s.upper())#upper case
print(s.lower())#lower case 
```
