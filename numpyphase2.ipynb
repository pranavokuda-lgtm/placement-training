numpy array operation

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50,60])
print("basic slicing:",arr[:4])
print("slicing with step:",arr[0:5:2])
print("negative indexing:",arr[-1])
```

```python
arr_2d=np.array([[10,20,30],
              [40,50,60],
              [70,80,90]])
#row,column
print("specific element:",arr_2d[1,2])
print("enteire row:",arr_2d[1:2])
```

```python
arr_1d=np.array([3,5,2,9,5,6,1,7])
print("sorting 1d array",sorted(arr_1d))
print("sorting 1d array",np.sort(arr_1d))
arr_2d=np.array([[2,1],
                [3,2],
                [4,3]])
#axis=0 for col axis=1 ->row
print("sort 2d array\n",np.sort(arr_2d,axis=0))
```

```python
number=np.array([3,5,2,9,5,6,1,7])
even_number=number[number%2==0]
print("Even number:",even_number)
```

```python
marks=np.array([67,34,10,23,50,97,54])
mask_pass=marks>35
print("More the 35(Mask):",mask_pass)
```

```python
sales=np.array([25000,12000,65000,80000])
idc=[1,3]
res=sales[idc]
print("Find value from index:",res)
```

```python
arr1=np.array([1,2,3,4])
arr2=np.array([10,20,30,40])
combined=np.concatenate([arr1,arr2])
print("Combined:",combined)
```

```python
matrix=np.array([[10,20],
                [30,40]])
newmatrix=np.array([[50,60],
                     [70,80]])
print("Join Horizantal:\n",np.hstack([matrix,newmatrix]))
print("Join Vertical:\n",np.vstack([matrix,newmatrix]))
```

```python

arr=np.array([1,2,3,4,5,6])
print("Split:",np.split(arr,2))
print("Append:",np.append(arr,50))
print("Insert:",np.insert(arr,0,90))
print("Remove",np.delete(arr,2))
print("Org:",arr)
```
